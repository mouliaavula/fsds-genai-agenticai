* Numpy vs pandas
* NumPy (Numerical Python) is a Python library mainly used for fast numerical and mathematical calculations.
* Its main data structure is an array (ndarray).
* Numpy array is a multidimentional , Un-lablled, Homogeneous data structure.
* But real time requiremnet is lablled data , hetrogeneous data and this is provided a by data structure called "Data Frames"
* Data frame supports only  two dimensional only
* The python library called Pandas/Polars suports working with Data Frames Data structure
* Pandas is a Python library mainly used for working with structured or tabular data, similar to an Excel table or database 
table.
* Data frame is a tabular structure  i.e table from that stores data as rows and 
* Data frame is a  2 dimensional labelled, hetrogeneous data structure used to store and work with data in rows and columns
* A Series is a 1-dimensional, labeled, hetrogeneous data structure. It is capable of holding any data type (integers, strings, 
* floating-point numbers, Python objects, etc.), but within a single Series, the data is usually homogeneous (all the same type).
* If you think of a DataFrame as an Excel spreadsheet, a Series is a single column from that spreadsheet
