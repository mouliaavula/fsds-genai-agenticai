```python
groups = ratings[['movieId','rating']].groupby('rating')
```

* Important: groupby() does NOT give you the final result yet
* groups:is a DataFrameGroupBy object.
* You're essentially saying: "Pandas, prepare these rows into groups based on rating."
* You haven't yet asked Pandas what to calculate for each group.
* Usually we perform an operation after groupby()
* groupby('rating').count()
* groupby('rating').mean()
* groupby('rating').sum()
* groupby('rating').max()
* groupby('rating').min()

```python
# with rating 0.5 there are 239125 movies are there
rating_counts = ratings[['movieId','rating']].groupby('rating').count()
rating_counts
```

```python
# For each rating value, count the non-null values in every remaining column.
# Why is rating not shown as a column?
# Because rating became the grouping index.
# If you want rating to remain a normal column
# ratings.groupby('rating').count().reset_index()
# rating → grouping key/index
# userId, movieId → columns on which count() is performed
# In previous query first we filtered columns to movieId and rating then applied grouping so rating column will become index column 
# and for movieId  count operations is applied where in below query directly applied grouping on rating so count operation is applied 
# on remaining columns. infact reaming all columns userId and movieId both shows same value
ratings.groupby('rating').count()
```

1. Split
   ↓
   Create groups based on rating

2. Apply
   ↓
   Calculate mean of movieId

3. Combine
   ↓
   Produce one result for each rating/group
