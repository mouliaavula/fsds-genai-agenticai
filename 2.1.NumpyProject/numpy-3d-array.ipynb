```python
import numpy as np
```

```python
image = np.array([
    [
        [255, 0, 0],      # Red
        [0, 255, 0],      # Green
        [0, 0, 255]       # Blue
    ],
    [
        [255, 255, 0],    # Yellow
        [255, 255, 255],  # White
        [0, 0, 0]         # Black
    ]
])
```

```python
image.shape
```

```python
image.ndim
```

There are 3 levels of [].

That's why this is a 3-dimensional NumPy array.

Dimension 1 → Row

Dimension 2 → Column

Dimension 3 → Value inside [255, 0, 0]

image[row, column, value]  to reach one individual number

(2,3,3)

2 → number of rows

3 → number of columns

3 → number of values in each element

Let's visualize the shape

Think of it as a table:

            Column 0       Column 1       Column 2


Row 0       [255,0,0]      [0,255,0]      [0,0,255]


Row 1       [255,255,0]    [255,255,255]  [0,0,0]

There are: 2 rows and 3 columns

And each cell contains:3 numbers

How to access elements?  image[row, column, position]

For a normal 2D array:  Important

array[row, column]

For your 3D array:

array[block, row, column]

For an RGB image, we'll eventually interpret the third position differently:

image[row, column, RGB_channel]

```python
#image[0,1] block 0 and row 2
image[0,1]
```

```python
#image[0,1] block 0 and row 3
image[0,2]
```

image[0,0] → [255, 0, 0]

image[0,1] → [0, 255, 0]

image[0,2] → [0, 0, 255]

Notice something important:

image[0,0] gives 3 numbers, not one number.

Why?

Because we have only specified:

block + row

We haven't specified which of the 3 values we want.

Access one individual number

Now we need the third index.

image[block, row, column] to compare with numpy array 

image[row(block), column(row), position(column)] to compare in terms of RGB



image = np.array([

                 col 1              col2            col3  

    row--> 0 [[255, 0, 0],     [0, 255, 0],     [0, 0, 255]], --one 2d array

    row ->1 [[255, 255, 0],   [255, 255, 255],   [0, 0, 0]] --another 2d array
])

[255,   0,   0] is  one pixel and is considered as [R,G,B] i.e R = 255 , G = 0 and B = 0

inside pixel R index is 0 and G index is 1 and Blue index 2


3d array and consists of 2 two dimensional arrays

here col1 each pixel/element has 3 elements [255,0,0]

```python
print(image[0,0,0]) # block 0 , row 0 , col 0
print(image[0,0,1]) # block 0 , row 0 , col 1
print(image[0,0,2]) # block 0 , row 0 , col 2
```
