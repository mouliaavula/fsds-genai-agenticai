* size()  → counts rows
* count() → counts non-NULL values

* What are the unique values?"-->unique() → give me the actual unique values

* How many unique values are there? --> nunique() → count unique values

* nunique() doesn't count NaN.

| Pandas      | SQL equivalent               | Meaning                         |
| ----------- | ---------------------------- | ------------------------------- |
| `size()`    | `COUNT(*)`                   | Count rows                      |
| `count()`   | `COUNT(column)`              | Count non-null values           |
| `nunique()` | `COUNT(DISTINCT column)`     | Count unique values             |
| `unique()`  | No direct `COUNT` equivalent | Return the actual unique values |


* size = rows, count = non-null, unique = values, nunique = number of unique value
