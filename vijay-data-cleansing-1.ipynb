```python
# marks above 100 replace with 0
df = pd.read_csv(r'E:\python-ds-gen-ai-vijay\data\eda\dirty_data2.csv')
df
```

```python
for i in df.index:
    if df.loc[i,'MATHS']>100:
        df.loc[i,'MATHS']= 0
df
```

```python
# we can also drop instead replace
df = pd.read_csv(r'E:\python-ds-gen-ai-vijay\data\eda\dirty_data2.csv')
for i in df.index:
    if df.loc[i,'MATHS']>100:
        df.drop(i,inplace=True)
df
```
