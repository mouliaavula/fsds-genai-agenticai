# Line Plot
* Use line plot when you want to answer "How something is changing". Most commonly used for time series data
* Example use cases: 
    * Stock price over time
    * Monthly sales
    * Temperature over days
    * Website visitors over months
* Trend ---> Line
* Use a Line Plot when you want to understand how data changes continuously, especially over time.
* Common questions a line plot can answer
* Is the value increasing or decreasing?
* What is the overall trend?
* When was the highest value?
* When was the lowest value?
* Is there any sudden increase or decrease?
* Is the data stable or fluctuating?
* Line plot conclusion = Understand the trend, direction, and changes in data over time.

```python
import matplotlib.pyplot as plt
```

```python
months = ['Jan','Feb','Mar','Apr','May','June']
sales = [100, 120, 140, 150, 180,200]

plt.plot(months, sales)
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Monthly Sales')
```

```python
months = ['Jan','Feb','Mar','Apr','May','June']
sales = [100, 120, 140, 150, 180,200]

plt.plot(months, sales)
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Monthly Sales')
plt.show()
```

* With above line plot;
    * we can say, sales are increasing in overal. i.e Increasing Trend from Jan to June
    * Min sale in Jan and Max sale in June
    * there was sharp fluctuation  from Mar to Apr and Apr to May

### Bar Chart — plt.bar()
* Use a bar chart when you want to compare different categories
* Use it for:
    * Sales by product
    * Employees by department
    * Population by city
    * Marks by subject
* Remember:Comparison → Bar
* Conclusions
    * Highest value:	Find the category with maximum value
    * Lowest value:	Find the category with minimum value
    * Compare categories :Compare values between categories	like Mobile sales are higher than Laptop
    * Difference between categories: See how much categories differ like Mobile sold 200 more than Laptop
    * Similar values: Identify categories with close values	Mobile and Laptop are relatively close
    * Ranking	Arrange categories from highest to lowest	Mobile → Laptop → Tablet
    * Unusual category	Identify unusually high or low values

```python
products = ['Mobile','Laptop','Tablet']
sales = [100,80,50]
plt.bar(products,sales)
plt.xlabel('Product')
plt.ylabel('Sales')
plt.show()
```

### Scatter Plot — plt.scatter()
* Use a scatter plot when you want to understand: "Is there a relationship between two numerical variables?"
* Remember:Relationship → Scatter
* Conclusion:
    * Points go upward : Positive relationship
    * Points go downward : Negative relationship
    * Points are random: No clear relationship
    * Points close together: Strong relationship
    * Points widely spread: Weak relationship
    * One unusual point: Outlier
    * Points form groups: Clusters
* Easy way to remember: Scatter Plot = Relationship between two numerical variables
* Ask these questions when looking at a scatter plot:
1. Is there a relationship?
2. Positive: if points generally move upward from left to right: As X increases, Y also increases.
3. negative:If points generally move downward from left to right:As X increases, Y decreases.
4. No relation:If points are randomly scattered:There is no clear relationship between X and Y.
5. Strong :Points are close together in a pattern:X and Y have a strong relationship.
6. weak:Points are more spread out:X and Y may have a weak relationship.
7. Are there any outliers:An outlier is a data point that is very different from other points.There may be an unusual observation or error in the data.
8. Are there any clusters:Sometimes points form groups.he data may contain different groups or categories.

```python
height = [150, 160, 165, 170, 180]
weight = [50, 55, 60, 65, 75]

plt.scatter(height, weight)
plt.xlabel("Height")
plt.ylabel("Weight")
plt.show()
```

* The above graph shows weak positive relation , no outlies and no clusters

### Histogram — plt.hist()
* Histogram = Understanding the distribution of numerical data. A histogram groups values into bins.

* When looking at a histogram, ask:
1. Where is most data concentrated?
2. What is the most frequent range?
3. How spread out is the data?
4. Is the data skewed?
5. Are there any unusual values?
6. Are there multiple peaks/groups?

* Important Histogram Patterns
1. Bell-shaped distribution:  Most values are concentrated around the middle.
2. Right-skewed distribution: Most values are low, but a few values are very high.
3. Left-skewed distribution: Most values are high, but a few values are very low.
4. Uniform distribution: Values are distributed almost equally across ranges.
5. Outlier: There may be an unusual value far away from most of the data.

* A bin is simply a range of values used to group numerical data.
* Think of bins as containers or buckets.
* plt.hist(ages, bins=4) means give data set divided into 4 groups or bins
* 20 - 25 bin 1 , 25-30 bin2, 30-35 bin3 and 35-40 bin4
* Bins = number of ranges/buckets used to group numerical data in a histogram.

* Minimum age = 20
* Maximum age = 40
* Range:40 - 20 = 20
* if bins = 4 
* Therefore, approximate width of each bin: 20/4 = range/number of bins= 5
* so bins are created like 20 - 25 , 25-30 , 30-35 and 35-40

[20, 25)
[25, 30)
[30, 35)
[35,40] # last bin includes max value

```python
ages = [20, 21, 22, 22, 23, 25, 25, 26, 28, 30, 35, 40]
plt.hist(ages,bins=8) # bin width = 40 -20 = 20/8 = 2.5
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

* [20 - 22.5) :20, 21, 22, 22 so frequency 4
* [22.5 - 25) :23 so frequency 1 
* [25 - 27.5) :25, 25, 26 so frequency 3
* [27.5 - 30) :28 so frequency 1
* [30 - 32.5) :30 so frequency 1
* [32.5 - 35) :no values so frequency 0
* [35 - 37.5) :35, so frequency 1 
* [37.7 - 40] :40 , so frequency 1

* Conclusion 1: Most ages are concentrated between 20 and 26
* Data in this range:20, 21, 22, 22, 23, 25, 25,26
* That's:8 out of 12 people
* So:Most people (about 66%) are between ages 20 and 26.

* Conclusion 2: The highest frequency age range is 20–22.5
* This bin contains:20,21,22,22
* Frequency:4
* Therefore:The most common age range in this dataset is approximately 20–22.5 years.

* Conclusion 3: The distribution is concentrated toward lower ages
* Most values are:20–30
* Only a few values are higher:35, 40
* The dataset is concentrated at younger ages, with fewer people at higher ages.

* Conclusion 4: There are gaps in the age distribution 32–34 and 36–38
* Some age ranges have no observations.

* Conclusion 5: The data is spread from 20 to 40: 40-20 = 20 years is wide spread

* Conclusion 6: The distribution is positively (right) skewed
* Most values are concentrated on the left:20 → 30
* A few values extend toward the right:35, 40
* Visually:
* More data                 Few data
* ███████████████               ██
* 20 ------------------------------> 40
* The distribution appears right-skewed (positively skewed), because most ages are lower and a few larger ages extend the distribution to the right.

###  Box Plot — plt.boxplot()
* Use a box plot when you want to understand:
* Median
* Minimum/maximum range
* Quartiles
* Outliers
* Remember:Outliers + spread → Box Plot
* A Box Plot (also called a Box-and-Whisker Plot) is mainly used to understand the distribution, spread, median, and outliers in numerical data.

* When looking at a Box Plot, ask these questions:
1. What is the median?
2. Where is the middle 50% of data?
3. How spread out is the data?
4. Are there any outliers?
5. Is the data symmetric or skewed?
6. How does one group compare with another?

Min        Q1          Median          Q3        Max
 |----------[=============|=============]----------|
            <------ IQR --------------->

```python
salary = [25, 30, 32, 35, 40, 42, 45, 100]
plt.boxplot(salary)
plt.ylabel('Salary')
plt.show()
```

```python
class_a = [60, 65, 70, 75, 80]
class_b = [50, 60, 70, 85, 90]

plt.boxplot([class_a, class_b])
plt.xticks([1, 2], ["Class A", "Class B"])
plt.ylabel("Marks")

plt.show()
```

### Stacked Bar Chart
* Use a stacked bar when you want to show:
* Category comparison + composition of each category.
* Remember:Comparison + composition → Stacked Bar

```python
departments = ["IT", "HR", "Sales"]

male = [50, 20, 40]
female = [30, 25, 35]

plt.bar(departments, male, label="Male")
plt.bar(departments, female, bottom=male, label="Female")

plt.legend()
plt.show()
```

### jointplot

* A Seaborn Joint Plot combines:
* Scatter plot (center) → relationship between two variables
* Distribution plot on X-axis → distribution of X variable
* Distribution plot on Y-axis → distribution of Y variable
* So, jointplot() helps us understand both relationship + distribution at the same time.

```python
import pandas as pd
```

```python
height = [150, 160, 165, 170, 180]
weight = [50, 55, 60, 65, 75]

df  = pd.DataFrame({"height":[150, 160, 165, 170, 180],"weight":[50, 55, 60, 65, 75]})
sns.jointplot(data=df,x='height',y='weight')
```

* Observation	Conclusion
* Scatter points move upward:	Positive relationship
* Points follow a straight pattern:	Approximately linear relationship
* Points are closely aligned:	Very strong correlation
* Height increases:	Weight generally increases
* No isolated points:	No obvious outliers
* Height distribution:	Mostly around 160–170
* Weight distribution:	Mostly around 50–65

### histplot() VS displot()

* histplot(): You already have a house and choose where to place a picture.
* displot():Seaborn builds the complete house for you.

* Use sns.histplot() when:
* You specifically want a histogram
* You want more control
* You are working with Matplotlib subplots
* You want to combine multiple graph types

* Use sns.displot() when:
* Your main goal is distribution analysis
* You want histogram/KDE/ECDF options
* You want to create separate plots by categories
* You want Seaborn to manage the complete figure


### SUB Plots : A subplot means putting multiple graphs inside one figure/window.

```python
# plt.subplot(rows, columns, position)
# 1 row 2 columns
plt.subplot(1,2,1)  # position 1
plt.plot([1,2,3],[10,20,30])
plt.subplot(1,2,2)# position 2
plt.plot(['A','B','C'],[40,50,60])
plt.show()
```

```python
plt.subplot(2,1,1)
plt.plot([1,2,3],[10,20,30])
plt.subplot(2,1,2)
plt.plot(['A','B','C'],[10,20,30])
plt.show()
```

```python
# 2 by 2 subplot

plt.subplot(2, 2, 1)
plt.plot([1, 2, 3], [10, 20, 30])

plt.subplot(2, 2, 2)
plt.bar(["A", "B", "C"], [10, 20, 15])

plt.subplot(2, 2, 3)
plt.scatter([1, 2, 3], [30, 20, 40])

plt.subplot(2, 2, 4)
plt.hist([10, 20, 20, 30, 30, 30])

plt.show()
```

### Modern and recommended way: plt.subplots()

```python
fig, ax = plt.subplots(1, 2)
ax[0].plot([1, 2, 3], [10, 20, 30])
ax[1].bar(["A", "B", "C"], [10, 20, 15])
plt.show()
```

```python
fig, ax = plt.subplots(2, 1)
ax[0].plot([1, 2, 3], [10, 20, 30])
ax[1].bar(["A", "B", "C"], [10, 20, 15])
plt.show()
```

```python
fig, ax = plt.subplots(2, 2)
ax[0, 0].plot([1, 2, 3], [10, 20, 30])
ax[0, 1].bar(["A", "B", "C"], [10, 20, 15])
ax[1, 0].scatter([1, 2, 3], [30, 20, 40])
ax[1, 1].hist([10, 20, 20, 30, 30, 30])
plt.show()
```

```python
fig, ax = plt.subplots(2, 2)

ax[0, 0].plot([1, 2, 3], [10, 20, 30])
ax[0, 0].set_title("Line Plot")

ax[0, 1].bar(["A", "B", "C"], [10, 20, 15])
ax[0, 1].set_title("Bar Plot")

ax[1, 0].scatter([1, 2, 3], [30, 20, 40])
ax[1, 0].set_title("Scatter Plot")

ax[1, 1].hist([10, 20, 20, 30, 30, 30])
ax[1, 1].set_title("Histogram")

plt.tight_layout()
plt.show()
```
