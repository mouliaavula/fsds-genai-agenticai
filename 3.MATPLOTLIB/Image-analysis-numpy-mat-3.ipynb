```python
# hourse_red array is a 3 dimensional array 
# 600 rows (image hieght)
# 1200 columns(image widht)
# 3--each element or cell or pixel has 3 colors Red, Green Blue--RGB. is also called as color channel
# and Red(is first color starts with 0)-->0 Green-->1 and Blue-->2

print(hourse_red[0,0])  # shows list with 3 elements[17 18 30]
print(hourse_red)
```

```python
plt.imshow(hourse_red[:,:,0])  # Take all rows, all columns, and only the Red channel.
```

```python
print(hourse_red[:,:,0].shape) # only red channel so it will become 2 dimensional
print(hourse_red[:,:,0])
print(hourse_red[:,:,0])
```

```python
plt.imshow(hourse_red[:,:,1])
print(hourse_red[:,:,1].shape) # only green channel so it will become 2 dimensional
```

```python
plt.imshow(hourse_red[:,:,2])
print(hourse_red[:,:,2].shape) # only blue channel so it will become 2 dimensional
```

```python
# by looking into all pics seems same but they are different . with below  mean it shows different value for each channel

print(hourse_red[:, :, 0].mean())
print(hourse_red[:, :, 1].mean())
print(hourse_red[:, :, 2].mean())
```

```python
# tried to visuvalize for 3 channels but we can not identify but they are different . matpotlib issue check chatgpt for more  info
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(hourse_red[:, :, 0], cmap='gray')
plt.title("Red Channel")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(hourse_red[:, :, 1], cmap='gray')
plt.title("Green Channel")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(hourse_red[:, :, 2], cmap='gray')
plt.title("Blue Channel")
plt.axis('off')

plt.show()
```
