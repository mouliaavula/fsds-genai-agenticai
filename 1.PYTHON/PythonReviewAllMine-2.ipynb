while loop -- always apply when you have condition 

```python
i = 1 #initiaize

while i<=5: # condition
    print("Data Science")
    i = i + 1 # increment
```

to understand nested while indepth understand you can use pycharm debug with f8 option

```python
i = 1
while i<=5:
    print("Data Science")
    j = 1
    while j<=4:
        print("Teachnology")
        j = j + 1
    i = i + 1        
```

```python
i = 1
while i<=5:
    print("Data Science", end='')
    j = 1
    while j<=4:
        print("Teachnology", end='')
        j = j+1
    i = i + 1
    print()
        
    
```

```python
# lets use while loop usig some numbers
i = 1

while i <= 2 :
    j = 0
    while  j <= 2 :
        print(i*j, end=" ")
        j += 1
    print()
    i += 1
```
