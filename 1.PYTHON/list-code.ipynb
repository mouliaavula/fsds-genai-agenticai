### all and any

```python
l = [1,2,3,4,0]
all(l) # returns true if all element s of list are greater than 0
```

```python
l = [1,2,3,4,5]
all(l)
```

```python
l = [1,2,3,4,5,True]
all(l)
```

```python
l = [1,2,3,4,5,True,False]
all(l)
```

```python
l = [1,2,3,4,0]
any(l) # returns true if any element of list are greater than 0
```

```python
l = [0]
any(l) # returns true if any element of list are greater than 0
```

```python
l = [1,True]
any(l) # returns true if any element of list are greater than 0
```

```python
l = [True]
any(l) # returns true if any element of list are greater than 0
```

```python
l = [False]
any(l) # returns true if any element of list are greater than 0
```

```python
l = [False,1]
any(l) # returns true if any element of list are greater than 0
```
