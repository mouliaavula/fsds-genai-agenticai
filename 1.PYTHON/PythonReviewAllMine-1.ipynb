### String slicing

default start index is  0  and  end index must be excluded i.e n-1

s[1:] - can you display from 1st index till last 

s[:5] - can you display from starting indext till n-1 | 4th 

s[2:7] - can you display from 2nd index to 7th ( n-1) 

s[2:10:2] - starting index 2, end index 10th and step - 2

When the step is positive , Python moves from left to right (forwward).

if step > 0 then start < stop  otherwise empty result  i.e start >= stop empty result

When the step is negative, Python moves from right to left (backwards).

if step < 0 then start > stop  otherwise empty result i.e start <= stop  empty result

```python
s = "Hello Python"
print(len(s))
print(s[::]) # total string
print(s[1:]) # second character to end of string
print(s[:5]) # first 5 characters
print(s[2:7]) # 3rd character to 7th 
print(s[7:]) # last 5 characters
print(s[:1]) # first character
print(s[-1:]) # last character
print(s[::-1]) # reverse of string
print(s[:13:2]) # with step 2
print(s[1:13:3]) # step 3
print(s[-1:-10:-2])
print(s[-4:-12:-3])
#print(s[-6:-1:-2])
```
