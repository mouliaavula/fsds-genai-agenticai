### Example of Pass by Reference
* Let's first see what true pass by reference means conceptually.
* Imagine a language that supports pass-by-reference:
* a → 10
* a ─────→ 10
* ↑
* │
* x
* then x = 100  inside function
* would effectively change the caller's variable:
* a → 100 and print(a) would give: 100
* Key point: With true pass-by-reference:
* The function can change what the caller's variable refers to.
* Python does not work this way

```python
# Example of Pass by Object Reference — Python
# Now let's see the actual Python behavior
def change(numbers):
    numbers.append(100)

my_list = [10, 20, 30]
change(my_list)
print(my_list)
```

my_list ────────>[10, 20, 30]<-------numbers

```python
# But look at this Python example carefully
# This is where beginners usually get confused.
def change(numbers):
    numbers = [100, 200, 300]

my_list = [10, 20, 30]
change(my_list)
print(my_list)
```

* Initially both point to my_list ────────>[10, 20, 30]<-------numbers
* But on numbers = [100, 200, 300]
* The parameter numbers is now pointing to a new object:
* my_list ─────→ [10, 20, 30]
* numbers ─────→ [100, 200, 300]
* my_list is not changed.

# Rule 1 — Pass by Value

A copy of the value is passed.

Function changes copy
        ↓
Original unchanged

# Rule 2 — Pass by Reference

A reference/alias to the caller's variable is passed.

Function reassigns parameter
        ↓
Caller variable can change

# Rule 3 — Python uses Object Reference

Python is best described as:

Pass by object reference / pass by assignment.

Not simply "pass by value" and not true "pass by reference."

# Rule 4 — Mutable objects can be modified

Examples:

list,dict,set

def change(x):

    x.append(100)

The original list can change.


# Rule 5 — Immutable objects cannot be modified

Common immutable objects:

int,float,str,tuple,bool

For example:

def change(x):

    x = 100

a = 10

change(a)

print(a)

Output:

10

# Rule 6 — Modification and reassignment are different

This is the most important Python rule.

Modification
x.append(100)

➡️ Changes the existing object.

Reassignment
x = [100, 200]

➡️ Makes x refer to a different object.

* Final Cheat Sheet

PASS BY VALUE

      ↓

Copy of value

      ↓

Changes don't affect original

PASS BY REFERENCE

      ↓

Reference to caller's VARIABLE

      ↓

Reassignment can affect caller

PYTHON: PASS BY OBJECT REFERENCE

      ↓
Reference to the OBJECT

      ↓

Mutable object → can modify original object

Immutable object → cannot modify it

      ↓

Reassignment of parameter → does NOT affect caller's variable

One sentence to remember for Python

Python passes object references. If the function mutates the object, the caller sees the change; 

if the function reassigns the parameter, the caller's variable is not changed.



### If an interviewer asks:

"Is Python pass by value or pass by reference?"

Don't simply answer "pass by value" or "pass by reference."

A better answer is:

"Python uses pass by object reference, also called pass by assignment. The function 

receives a reference to the same object. If the object is mutable, we can modify the 

original object and if object immutable then we can not modify original object. If we 

reassign the parameter, it only changes what the local parameter refers to."

```python
def test(x):
    print(id(x))       # Same as id(a)

    x.append(100)

    print(id(x))       # Still same as id(a)

    x = [500, 600]     # Reassignment  difference between pass by reference and pass by object reference  comes at this point. 
    
    #  pass by reference caller can see difference with [500,600] ie. a is updated to [500,600] 
    #  and pass by object reference caller can't see  i.e both a nd x are different

    print(id(x))       # NOW different from id(a)


a = [10, 20, 30]

print(id(a))

test(a)

print(id(a))
print(a)
```

- KWARGs (key worded variable length arguments)

```python
def person(name,*data):
    print(name)
    print(data)

person('ALEX', 36, 'JOHN', 987767)
```

```python
def person(name,*data):
    print('name')
    print(data)

person('ALEX', age = 36, home_place ='southcity', mob =987767)
# we got error as keyword argument thats why we add another *
```

```python
def person(name, **data):
    print(name)
    print(data)

person('mark', age = 36, home_place ='southcity', mob =987767)
```

```python
def person(name, **data):
    print(name)
    
    for i, j in data.items():
        print(i, j)
        
person('john', age = 36, home_place ='southcity', mob =987767, place = 'USA')
```
